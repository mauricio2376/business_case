```python
# ===============================================
# Notebook: Load Staging Sales via Unity Catalog
# ===============================================

from pyspark.sql.functions import col

# --------------------------------------
# 1. Leitura do CSV de vendas no DBFS
# --------------------------------------
filename_sales = "/FileStore/bronze/abi_bus_case1_beverage_channel_group_20210726.csv"

df = spark.read \
    .option("header", True) \
    .option("encoding", "utf-8") \
    .option("sep", ",") \
    .csv(filename_sales)

# --------------------------------------
# 2. Variáveis para escrita
# --------------------------------------
catalog = "beverage_analytics"
schema = "staging"
table = "abi_bus_case1_beverage_channel_group_20210726"

# --------------------------------------
# 3. Escrita no Unity Catalog (Delta Table)
# --------------------------------------
df.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable(f"{catalog}.{schema}.{table}")

print(f"Carga concluída com sucesso em: {catalog}.{schema}.{table}")
```
